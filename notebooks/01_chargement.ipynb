# 01 - Chargement du dataset Online Retail II

**Objectif du notebook :** charger le dataset brut, vérifier son intégrité (structure, types, volume) et sauvegarder un snapshot exploitable pour la suite du projet.

**Cadrage, Acquisition & Nettoyage**


## 2. Chargement des deux feuilles Excel et fusion

```python
import pandas as pd

df_2009 = pd.read_excel("../data/raw/online_retail_II.xlsx", sheet_name="Year 2009-2010")
df_2010 = pd.read_excel("../data/raw/online_retail_II.xlsx", sheet_name="Year 2010-2011")

df = pd.concat([df_2009, df_2010], ignore_index=True)

print(df.shape)
df.head()
```

## 3. Vérifications de base

```python
# Types de colonnes
df.dtypes

```

```python
# Vérifier que les colonnes attendues sont présentes
df.columns.tolist()
# Attendu : ['Invoice', 'StockCode', 'Description', 'Quantity', 
#            'InvoiceDate', 'Price', 'Customer ID', 'Country']

```

```python
# Volume total et periode couverte
print("Nombre total de lignes :", len(df))
print("Periode couverte : du", df["InvoiceDate"].min(), "au", df["InvoiceDate"].max())
```

```python
# Vérifier que les dates sont bien reconnues comme datetime
df["InvoiceDate"].dtype
```

```python
# filtre le DataFrame pour afficher les cinq premiers numéros de facture (Invoice) qui commencent par la lettre C,
# correspondant aux factures d'annulation (retours) dans le dataset Online Retail II.
df[df["Invoice"].astype(str).str.startswith("C")]["Invoice"].head()
```

```python
# sélectionne toutes les colonnes de type object du DataFrame et les convertit en type string afin d'uniformiser 
#les types de données et d'éviter des erreurs lors de l'export (par exemple vers Parquet).

object_cols = df.select_dtypes(include="object").columns

for col in object_cols:
    df[col] = df[col].astype("string")
```

## 4. Sauvegarde d'un snapshot (optionnel mais recommandé)

Permet de recharger rapidement le dataset dans les prochains notebooks sans repasser par l'Excel (lent).

```python
# Pour ne pas recharger l'Excel (lent) à chaque fois , Ensuite on pourra recharger rapidement avec pd.read_parquet(...)
df.to_parquet("../data/raw/online_retail_full.parquet", index=False)
```
