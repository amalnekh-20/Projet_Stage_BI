```python
import sys
sys.path.append("../scripts")
from clean_data import load_raw_data, clean_data

df_raw = load_raw_data()
print("Avant nettoyage :", df_raw.shape)

resultats = clean_data(df_raw)

for nom, df in resultats.items():
    print(f"{nom} : {df.shape}")
```
